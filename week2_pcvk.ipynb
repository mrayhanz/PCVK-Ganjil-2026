```python
canvas_int16 = np.zeros(shape=(530,530,3), dtype=np.int16)
canvas_int32 = np.zeros(shape=(530,530,3), dtype=np.int32)

print("int16")
print("dtype :", canvas_int16.dtype)
print("min   :", canvas_int16.min())
print("max   :", canvas_int16.max())
print("nbytes:", canvas_int16.nbytes)

print("\nint32")
print("dtype :", canvas_int32.dtype)
print("min   :", canvas_int32.min())
print("max   :", canvas_int32.max())
print("nbytes:", canvas_int32.nbytes)

plt.subplot(1,2,1)
plt.imshow(canvas_int16)
plt.title("int16")

plt.subplot(1,2,2)
plt.imshow(canvas_int32)
plt.title("int32")

plt.show()
```

Tampilan keduanya sama-sama hitam karena nilai pikselnya 0. Memorinya berbeda karena int16 menggunakan 2 byte, sedangkan int32 menggunakan 4 byte, sehingga int32 membutuhkan memori 2× lebih besar.

```python
def label_identitas(img,nama):

    h,w = img.shape[:2]

    bar_height = int(h * 0.10)

    cv.rectangle(
        img,
        pt1=(0,h-bar_height),
        pt2=(w,h),
        color=(0,0,0),
        thickness=-1
    )

    font = cv.FONT_HERSHEY_SIMPLEX

    font_scale = min(w,h) / 500
    thickness = max(1,int(min(w,h) / 250))

    cv.putText(
        img,
        text=nama,
        org=(int(w*0.03),h-int(bar_height*0.35)),
        fontFace=font,
        fontScale=font_scale,
        color=(255,255,255),
        thickness=thickness,
        lineType=cv.LINE_AA
    )

    return img
```

```python
img = cv.imread('aktivitas.jpeg')

hasil = label_identitas(
    img.copy(),
    'Muhammad Rayhan'
)

plt.imshow(cv.cvtColor(hasil,cv.COLOR_BGR2RGB))
plt.show()
```

```python
img1 = cv.imread('aktivitas.jpeg')
img2 = cv.imread('aktivitas1.jpeg')

hasil1 = label_identitas(img1.copy(),'rayy')
hasil2 = label_identitas(img2.copy(),'rayy')

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(hasil1,cv.COLOR_BGR2RGB))
plt.title('Foto 1')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(hasil2,cv.COLOR_BGR2RGB))
plt.title('Foto 2')

plt.show()
```

```python
def thumbnail_1_1(img,nama_toko):

    h,w = img.shape[:2]

    sisi = max(h,w)

    canvas = np.full(
        (sisi,sisi,3),
        255,
        dtype=np.uint8
    )

    scale = min(sisi/w,sisi/h)

    new_w = int(w*scale)
    new_h = int(h*scale)

    img_resize = cv.resize(
        img,
        (new_w,new_h)
    )

    x = (sisi-new_w)//2
    y = (sisi-new_h)//2

    canvas[
        y:y+new_h,
        x:x+new_w
    ] = img_resize

    font = cv.FONT_HERSHEY_SIMPLEX

    font_scale = sisi/1000
    thickness = max(1,int(sisi/300))

    cv.putText(
        canvas,
        text=nama_toko,
        org=(int(sisi*0.03),int(sisi*0.95)),
        fontFace=font,
        fontScale=font_scale,
        color=(0,0,0),
        thickness=thickness,
        lineType=cv.LINE_AA
    )

    return canvas
```

```python
img = cv.imread('produk.jpeg')

hasil = thumbnail_1_1(
    img,
    'SUSU PASTEURISASI'
)

plt.figure(figsize=(8,8))
plt.imshow(cv.cvtColor(hasil,cv.COLOR_BGR2RGB))
plt.title('Thumbnail 1:1')
plt.show()

print("Ukuran thumbnail :",hasil.shape)
```
